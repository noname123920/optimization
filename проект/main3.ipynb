```octave
dt = 0.1;
T = 5;
u = 2;

u_history = [u];
t_history = [0];

n = 1;
while n * dt <= T
    if u > 0
        f = -1;
    else
        f = 1;
    endif
    
    u = u + dt * f;
    
    u_history = [u_history, u];
    t_history = [t_history, n * dt];
    
    n = n + 1;
endwhile

plot(t_history, u_history, 'b-', 'LineWidth', 4);
grid on;
xlabel('t, c');
ylabel('u(t)');
```
