```octave
graphics_toolkit('gnuplot');
```

```octave
f = @(x) 1 ./ (1 + 25 * x.^2);
x = linspace(-5, 5, 1000);

n = 50;

xi = linspace(-5, 5, n+1);
yi = f(xi);

y_interp = zeros(size(x));
for i = 1:n+1
    L = ones(size(x));
    for j = 1:n+1
        if j ~= i
            L = L .* (x - xi(j)) / (xi(i) - xi(j));
        end
    end
    y_interp = y_interp + yi(i) * L;
end

plot(x, f(x), 'b-', 'LineWidth', 6); 
hold on;
plot(x, y_interp, 'r-', 'LineWidth', 6);
plot(xi, yi, 'ro', 'MarkerSize', 8, 'MarkerFaceColor', 'k', 'LineWidth', 1.5);
hold off;

grid on;
xlabel('x');
ylabel('y');
xlim([-5, 5]);
ylim([-0.5, 2]);
```
